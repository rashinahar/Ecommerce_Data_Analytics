```python
#User 2

#17:51 → view
#17:53 → view       ← same session
#18:01 → view       ← same session
#18:20 → view       ← same session
#19:05 → view       ← 45 min gap → NEW session
#if gap between two consecutive event is greater than or eequal to 30
#min than new session will be started
time_diff = events.groupby("visitorid")["timestamp"].diff()

events["new_session"] = (
    time_diff.isnull() | (time_diff > pd.Timedelta(minutes=30))
)

print(
    events[
        ["visitorid", "timestamp", "event", "new_session"]
    ].head(15)
)
```

same session(unordered) view → addtocart → transaction

```python
def check_funnel(events):
    events = set(events)

    return (
        "view" in events
        and "addtocart" in events
        and "transaction" in events
    )

complete_funnel_sessions = events.groupby("session_key")["event"].apply(
    check_funnel
)

print(
    "Sessions with complete funnel:",
    complete_funnel_sessions.sum()
)
```

sequence-based funnel

```python
def check_ordered_funnel(group):
    events_list = group["event"].tolist()

    try:
        view_index = events_list.index("view")
        cart_index = events_list.index("addtocart")
        transaction_index = events_list.index("transaction")

        return (
            view_index < cart_index < transaction_index
        )

    except ValueError:
        return False


ordered_funnel = events.groupby("session_key").apply(
    check_ordered_funnel
)

print(
    "Sessions with ordered View → Cart → Purchase:",
    ordered_funnel.sum()
)
```

Calculation on the basis of ordered funnel

```python
session_event_flags = events.pivot_table(
    index="session_key",
    columns="event",
    values="visitorid",
    aggfunc="size",
    fill_value=0
)

session_event_flags["has_view"] = session_event_flags["view"] > 0
session_event_flags["has_cart"] = session_event_flags["addtocart"] > 0
session_event_flags["has_purchase"] = session_event_flags["transaction"] > 0

print(session_event_flags.head())
```

```python
total_sessions = len(session_event_flags)

view_sessions = session_event_flags["has_view"].sum()
cart_sessions = session_event_flags["has_cart"].sum()
purchase_sessions = session_event_flags["has_purchase"].sum()

view_cart_sessions = (
    session_event_flags["has_view"] &
    session_event_flags["has_cart"]
).sum()

cart_purchase_sessions = (
    session_event_flags["has_cart"] &
    session_event_flags["has_purchase"]
).sum()

all_three_sessions = (
    session_event_flags["has_view"] &
    session_event_flags["has_cart"] &
    session_event_flags["has_purchase"]
).sum()

print("Total Sessions:", total_sessions)
print("View Sessions:", view_sessions)
print("Cart Sessions:", cart_sessions)
print("Purchase Sessions:", purchase_sessions)
print("View + Cart:", view_cart_sessions)
print("Cart + Purchase:", cart_purchase_sessions)
print("View + Cart + Purchase:", all_three_sessions)

print("\n--- Funnel Rates ---")
print("View → Cart:", round(view_cart_sessions / view_sessions * 100, 2), "%")
print("Cart → Purchase:", round(cart_purchase_sessions / cart_sessions * 100, 2), "%")
print("View → Purchase:", round(all_three_sessions / view_sessions * 100, 2), "%")
```

```python
session_order = events.pivot_table(
    index="session_key",
    columns="event",
    values="timestamp",
    aggfunc="min"
)

session_order["view_before_cart"] = (
    session_order["view"] < session_order["addtocart"]
)

session_order["cart_before_purchase"] = (
    session_order["addtocart"] < session_order["transaction"]
)

ordered_funnel_sessions = (
    session_order["view_before_cart"] &
    session_order["cart_before_purchase"]
).sum()

print("Ordered View → Cart → Purchase Sessions:", ordered_funnel_sessions)
```

```python
ordered_conversion_rate = (
    ordered_funnel_sessions / session_order["view"].notna().sum()
) * 100

print(
    "Strict View → Cart → Purchase Conversion Rate:",
    round(ordered_conversion_rate, 2),
    "%"
)
```

```python
product_events = events[
    ["session_key", "itemid", "event", "timestamp"]
].copy()

product_events["view_time"] = product_events["timestamp"].where(
    product_events["event"] == "view"
)

product_events["cart_time"] = product_events["timestamp"].where(
    product_events["event"] == "addtocart"
)

product_events["purchase_time"] = product_events["timestamp"].where(
    product_events["event"] == "transaction"
)

product_session_summary = product_events.groupby(
    ["session_key", "itemid"]
)[["view_time", "cart_time", "purchase_time"]].min().reset_index()

print(product_session_summary.head())
```

```python
product_session_summary["ordered_view_cart_purchase"] = (
    product_session_summary["view_time"].notna()
    & product_session_summary["cart_time"].notna()
    & product_session_summary["purchase_time"].notna()
    & (product_session_summary["view_time"] < product_session_summary["cart_time"])
    & (product_session_summary["cart_time"] < product_session_summary["purchase_time"])
)

print(
    product_session_summary[
        product_session_summary["ordered_view_cart_purchase"]
    ].head()
)
```

```python
product_funnel = product_session_summary.groupby("itemid").agg(
    views=("view_time", "count"),
    carts=("cart_time", "count"),
    ordered_purchases=("ordered_view_cart_purchase", "sum")
).reset_index()

print(product_funnel.head())
```

```python
product_funnel["view_to_cart_rate"] = (
    product_funnel["carts"]
    / product_funnel["views"]
    * 100
)

product_funnel["view_to_purchase_rate"] = (
    product_funnel["ordered_purchases"]
    / product_funnel["views"]
    * 100
)

print(product_funnel.head())
```

A minimum threshold of 100 views is used to avoid misleading conversion rates from products with very few views. For example, a product with only 2 views and 1 purchase would have a 50% conversion rate, but this does not provide enough data to conclude that the product is actually a high-converting product.


```python
category_event_summary["view_to_cart_rate"] = (
    category_event_summary["cart_adds"]
    / category_event_summary["views"]
    * 100
)

category_event_summary["cart_to_purchase_rate"] = (
    category_event_summary["purchases"]
    / category_event_summary["cart_adds"]
    * 100
)

category_event_summary["view_to_purchase_rate"] = (
    category_event_summary["purchases"]
    / category_event_summary["views"]
    * 100
)

print(category_event_summary.head(10))
```

Cohort / Retention Analysis

```python
user_first_month = events.groupby("visitorid")["timestamp"].min().reset_index()

user_first_month["cohort_month"] = (
    user_first_month["timestamp"].dt.to_period("M")
)

print(user_first_month.head())
```

```python
user_activity = events[["visitorid", "year_month"]].drop_duplicates()

user_activity = user_activity.merge(
    user_first_month[["visitorid", "cohort_month"]],
    on="visitorid",
    how="left"
)

print(user_activity.head(10))
```

```python
user_activity["cohort_index"] = (
    user_activity["year_month"].dt.year
    - user_activity["cohort_month"].dt.year
) * 12 + (
    user_activity["year_month"].dt.month
    - user_activity["cohort_month"].dt.month
)

print(user_activity.head(10))
```

```python
cohort_retention = user_activity.groupby(
    ["cohort_month", "cohort_index"]
)["visitorid"].nunique().reset_index()

cohort_retention = cohort_retention.pivot(
    index="cohort_month",
    columns="cohort_index",
    values="visitorid"
)

print(cohort_retention)
```

```python
cohort_retention_pct = cohort_retention.div(
    cohort_retention[0],
    axis=0
) * 100

print(cohort_retention_pct.round(2))
```

heatmap
